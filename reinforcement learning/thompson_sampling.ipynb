```BENEFITS```
- Each machine starts with a random chance of being the best
- Every time you play, you update waht you know about each machine
- Instead of just picking the machine with the highest average, you randomly choose based on how confident you are that a machine is the best 

```Why is this better:```- Even if a machine was lucky early, Thompson Sampling still tests other machines in a smart way. It doesn't get stuck on a bad choice


````python
import random
````

````python
n_machines = 3
n_trials = 10
````

````python
true_rewards = [0.2, 0.5, 0.8]
````

````python
success = [0] * n_machines
failures = [0] * n_machines
````

````python
for i in range(n_trials):
    sampled_values = [random.betavariate(success[i] + 1, failures[i] + 1) for i in range(n_machines)]
    arm = sampled_values.index(max(sampled_values)) # choose the best guess
    reward = 1 if random.random() < true_rewards[arm] else 0
    if reward:
        success[arm] += 1
    else:
        failures[arm] += 1
````

````python
for i in range(n_machines):
    total = success[i] + failures[i]
    print(f"Machine {i+1}; Played {total} times; Estimated win rate: {success[i] / total:.2f}")

print("Best Machine to Play:", success.index(max(success)) + 1)
````
