```BENEFITS```
- Try each machine one in the beginning
- After that, pick the machine that has the best average score
- If two machines seem close, give more chances to the one that hasn't bee tested much 

```PROBLEM```
- If one machine was lucky in teh first few rounds, UCB might keep picking it, even if another machine is actually better


````python
import math
import random 
````

````python
n_machines = 3
n_trials = 10
````

- count[i] stores how many times machine i has been played
- rewards[i] stores how many total rewards were received from machine i 
- initially, all values are 0 because we have not played yet


````python
count =[0] * n_machines
rewards = [0] * n_machines
````

````python
true_rewards = [0.2, 0.5, 0.8]
````

````python
# This loop runs 10 times(since n_trials = 10)
# each time, agent chooses a slot machine machine to play
for t in range(1,n_trials+1):
    # If any machine has not been played yet (0 in count), we choose it first 
    # This ensures each machine is played at teast once before applying the UCB formula, this is called exploration
    if 0 in count:
        arm = count.index(0)
    else:
        # Calculate UCB for each machine 
        #average reward + exploration bonus
        ucb_values = [
            rewards[i] / count[i] + math.sqrt(2 * math.log(t) / count[i]) 
            for i in range(n_machines)
        ]
        arm = ucb_values.index(max(ucb_values)) # pick the machine with highest UCB value

    reward = 1 if random.random() < true_rewards[arm] else 0
    count[arm] += 1
    rewards[arm] += reward 
````

````python
# show result
for i in range(n_machines):
    print(f"Machine {i+1}; Played {count[i]} times; Estimated win rate:{rewards[i]/count[i]:.2f}") 

print("Best Machine to Play:", rewards.index(max(rewards))+1)
````
